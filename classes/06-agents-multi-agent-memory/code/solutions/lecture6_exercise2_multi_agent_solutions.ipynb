## Part A — A supervisor with specialized agents (slides 32–34)
An agent does *not* automatically become a domain expert because it has a different name; here the specialists differ in instructions and responsibilities. This exercise uses simulated research rather than live web browsing.

```python
class TeamState(TypedDict):
    task: str
    route: str
    output: str

class RouteDecision(BaseModel):
    next_agent: Literal["researcher", "writer"]

supervisor_model_e2 = llm.with_structured_output(RouteDecision)

def supervisor_e2(state: TeamState):
    decision = supervisor_model_e2.invoke([
        ("system", "Route information-gathering requests to researcher and drafting requests to writer. Output only the structured choice."),
        ("human", state["task"]),
    ])
    return {"route": decision.next_agent}

def researcher_e2(state: TeamState):
    # No live search: findings are a simulated reference.
    return {"output": "SIMULATED RESEARCH: LangGraph uses nodes, edges and state to coordinate execution."}

def writer_e2(state: TeamState):
    answer = llm.invoke([
        ("system", "Write a short, accessible answer. Don't claim to have searched the web."),
        ("human", state["task"]),
    ])
    return {"output": str(answer.content)}

def route_team_e2(state: TeamState) -> Literal["researcher", "writer"]:
    return state["route"]
```

### Build the supervisor graph
The supervisor routes each task to the appropriate specialist. Read the conditional edge mapping and test both routes.

```python
builder_e2 = StateGraph(TeamState)
builder_e2.add_node("supervisor", supervisor_e2)
builder_e2.add_node("researcher", researcher_e2)
builder_e2.add_node("writer", writer_e2)
builder_e2.add_edge(START, "supervisor")
builder_e2.add_conditional_edges("supervisor", route_team_e2, {
    "researcher": "researcher", "writer": "writer"
})
builder_e2.add_edge("researcher", END)
builder_e2.add_edge("writer", END)
app_e2 = builder_e2.compile()

for task in ["Gather facts about LangGraph state", "Draft a friendly introduction to Python dictionaries"]:
    result = app_e2.invoke({"task": task, "route": "", "output": ""})
    print("Route:", result["route"])
    print("Output:", result["output"])
```

### Solution: Add a reviewer specialist
Create a new structured routing schema and rebuild the graph. The reviewer should receive a draft to critique in its task description.

```python
class RouteDecision3(BaseModel):
    next_agent: Literal["researcher", "writer", "reviewer"]

supervisor_model_e2_three = llm.with_structured_output(RouteDecision3)

def supervisor_three_e2(state: TeamState):
    choice = supervisor_model_e2_three.invoke([
        ("system", "Choose researcher for fact gathering, writer for drafting, "
                   "or reviewer for critique. Return the structured selection."),
        ("human", state["task"]),
    ])
    return {"route": choice.next_agent}

def reviewer_e2(state: TeamState):
    response = llm.invoke([
        ("system", "Review the supplied text. Give one constructive, concrete suggestion. "
                   "Do not invent external verification."),
        ("human", state["task"]),
    ])
    return {"output": str(response.content)}

builder_e2_three = StateGraph(TeamState)
builder_e2_three.add_node("supervisor", supervisor_three_e2)
builder_e2_three.add_node("researcher", researcher_e2)
builder_e2_three.add_node("writer", writer_e2)
builder_e2_three.add_node("reviewer", reviewer_e2)
builder_e2_three.add_edge(START, "supervisor")
builder_e2_three.add_conditional_edges("supervisor", route_team_e2, {
    "researcher": "researcher", "writer": "writer", "reviewer": "reviewer"
})
for specialist in ("researcher", "writer", "reviewer"):
    builder_e2_three.add_edge(specialist, END)
app_e2_three = builder_e2_three.compile()

review_request = "Review this draft for clarity: Python dictionaries store key-value pairs."
result = app_e2_three.invoke({"task": review_request, "route": "", "output": ""})
print("Selected:", result["route"])
print("Result:", result["output"])
```

## Part B — Sequential collaboration (slides 36–40)
This is a different pattern: steps are predefined, so no supervisor LLM is needed to choose the route. Compare it with Part A.

```python
class DraftReviewState(TypedDict):
    request: str
    draft: str
    feedback: str

def draft_node(state: DraftReviewState):
    result = llm.invoke([("system", "Write a very short introductory paragraph."), ("human", state["request"])])
    return {"draft": str(result.content)}

def review_node(state: DraftReviewState):
    result = llm.invoke([("system", "Review the draft. Give one specific improvement in one sentence."), ("human", state["draft"])])
    return {"feedback": str(result.content)}

pipeline = StateGraph(DraftReviewState)
pipeline.add_node("writer", draft_node)
pipeline.add_node("reviewer", review_node)
pipeline.add_edge(START, "writer")
pipeline.add_edge("writer", "reviewer")
pipeline.add_edge("reviewer", END)
pipeline_app = pipeline.compile()

out = pipeline_app.invoke({"request": "Introduce Python dictionaries to beginners", "draft": "", "feedback": ""})
print("DRAFT:\n", out["draft"])
print("REVIEW:\n", out["feedback"])
```

### Discussion
- When do we need a supervisor versus a fixed pipeline?
- What would happen if the Writer gives incorrect facts to the Reviewer?
- Which stages could be deterministic Python functions instead of LLM agents?

⭐ **Stretch:** Let the reviewer trigger one revision, with a maximum of two iterations.

**Success:** Students can build a supervisor and contrast dynamic routing with fixed sequential handoffs.

**Instructor answer:** A supervisor is useful when routing is conditional and depends on task intent. A pipeline is simpler when the order is always fixed. Both require error handling; a reviewer can repeat a writer’s misinformation rather than catch it.

### ⭐ Optional solution: Revision after feedback
This graph adds a single revision step. We avoid an unbounded reviewer–writer loop.

```python
class RevisionState(TypedDict):
    request: str
    draft: str
    feedback: str
    revised: str

def make_draft(state: RevisionState):
    response = llm.invoke([
        ("system", "Write a short explanation for Python beginners."),
        ("human", state["request"]),
    ])
    return {"draft": str(response.content)}

def critique_draft(state: RevisionState):
    response = llm.invoke([
        ("system", "Identify one concrete improvement in this explanation."),
        ("human", state["draft"]),
    ])
    return {"feedback": str(response.content)}

def revise_once(state: RevisionState):
    response = llm.invoke([
        ("system", "Revise the draft once using the feedback."),
        ("human", f"Draft: {state['draft']}\nFeedback: {state['feedback']}")
    ])
    return {"revised": str(response.content)}

revision_builder = StateGraph(RevisionState)
revision_builder.add_node("writer", make_draft)
revision_builder.add_node("reviewer", critique_draft)
revision_builder.add_node("revision", revise_once)
revision_builder.add_edge(START, "writer")
revision_builder.add_edge("writer", "reviewer")
revision_builder.add_edge("reviewer", "revision")
revision_builder.add_edge("revision", END)
revision_app = revision_builder.compile()
revision_result = revision_app.invoke({
    "request": "Explain a Python dictionary with one example.",
    "draft": "", "feedback": "", "revised": "",
})
print("FIRST DRAFT:\n", revision_result["draft"])
print("FEEDBACK:\n", revision_result["feedback"])
print("REVISED:\n", revision_result["revised"])
```
