### A0 — basic data types

```python
name = "Fatima Al-Rashid"
date_of_birth = "1990-04-12"  # str — no special "date" type without parsing
age = 34
is_paid_member = True
membership_price = 12.50

print(type(name))              # str
print(type(date_of_birth))     # str
print(type(age))               # int
print(type(is_paid_member))    # bool
print(type(membership_price))  # float

assert isinstance(name, str)
assert isinstance(date_of_birth, str)
assert isinstance(age, int) and not isinstance(age, bool)
assert isinstance(is_paid_member, bool)
assert isinstance(membership_price, float)
print("A0 passed!")
```

### A1 — classify_temperature

```python
def classify_temperature(celsius):
    if celsius <= 0:
        return "freezing"
    elif celsius <= 15:
        return "cold"
    elif celsius <= 25:
        return "mild"
    else:
        return "hot"

assert classify_temperature(-5) == "freezing"
assert classify_temperature(10) == "cold"
assert classify_temperature(20) == "mild"
assert classify_temperature(30) == "hot"
assert classify_temperature(0) == "freezing"
assert classify_temperature(15) == "cold"
print("A1 passed!")
```

### A2 — build_greeting

```python
def build_greeting(name, greeting="Hello", **details):
    base = f"{greeting}, {name}!"
    if details:
        extras = ", ".join(f"{k}: {v}" for k, v in details.items())
        return f"{base} ({extras})"
    return base

assert build_greeting("Silvia", role="data steward", city="Aarhus") == \
    "Hello, Silvia! (role: data steward, city: Aarhus)"
assert build_greeting("Azim") == "Hello, Azim!"
assert build_greeting("Azim", greeting="Hi") == "Hi, Azim!"
print("A2 passed!")
```

### A3 — pending_tasks_by_priority

```python
tasks = [
    {"name": "read paper", "priority": 2, "done": False},
    {"name": "write code", "priority": 1, "done": False},
    {"name": "email advisor", "priority": 3, "done": True},
    {"name": "prepare slides", "priority": 1, "done": False},
]

def pending_tasks_by_priority(tasks):
    pending = [t for t in tasks if not t["done"]]
    pending_sorted = sorted(pending, key=lambda t: t["priority"])
    return [t["name"] for t in pending_sorted]

result = pending_tasks_by_priority(tasks)
assert result == ["write code", "prepare slides", "read paper"], result
print("A3 passed!")
```

### A4 — count_by_priority

```python
def count_by_priority(tasks):
    # one-liner version
    priorities = {t["priority"] for t in tasks}
    return {p: len([t for t in tasks if t["priority"] == p]) for p in priorities}

def count_by_priority_longhand(tasks):
    counts = {}
    for t in tasks:
        p = t["priority"]
        if p not in counts:
            counts[p] = 0
        counts[p] += 1
    return counts

result = count_by_priority(tasks)
assert result == {2: 1, 1: 2, 3: 1}, result
assert count_by_priority_longhand(tasks) == result
print("A4 passed!")
```

### A5 — safe_call

```python
def flaky_divide(a, b):
    return a / b

def safe_call(func, a, b):
    try:
        return func(a, b)
    except ZeroDivisionError:
        return "error: division by zero"
    except Exception as e:
        return f"error: {type(e).__name__}"

assert safe_call(flaky_divide, 10, 2) == 5.0
assert safe_call(flaky_divide, 10, 0) == "error: division by zero"
assert safe_call(flaky_divide, 10, "a") == "error: TypeError"
print("A5 passed!")
```

### B1 — Task class

```python
class Task:
    def __init__(self, name, priority):
        self.name = name
        self.priority = priority
        self.done = False

    def mark_done(self):
        self.done = True

    def __str__(self):
        box = "[x]" if self.done else "[ ]"
        return f"{box} {self.name} (priority {self.priority})"

t = Task("write code", 1)
assert t.name == "write code"
assert t.priority == 1
assert t.done is False
assert str(t) == "[ ] write code (priority 1)"
t.mark_done()
assert t.done is True
assert str(t) == "[x] write code (priority 1)"
print("B1 passed!")
```

### B2 — class attribute total_created

```python
class Task:
    total_created = 0

    def __init__(self, name, priority):
        self.name = name
        self.priority = priority
        self.done = False
        Task.total_created += 1

    def mark_done(self):
        self.done = True

    def __str__(self):
        box = "[x]" if self.done else "[ ]"
        return f"{box} {self.name} (priority {self.priority})"

Task.total_created = 0
t1 = Task("a", 1)
t2 = Task("b", 2)
t3 = Task("c", 3)
assert Task.total_created == 3
assert t1.total_created == 3
print("B2 passed!")
```

### B3 — TaskList

```python
class TaskList:
    def __init__(self):
        self.tasks = []

    def add(self, task):
        self.tasks.append(task)

    def pending(self):
        pending = [t for t in self.tasks if not t.done]
        return sorted(pending, key=lambda t: t.priority)

    def __len__(self):
        return len(self.tasks)

tl = TaskList()
tl.add(Task("write code", 1))
tl.add(Task("read paper", 2))
tl.add(Task("email advisor", 3))
assert len(tl) == 3
pending_names = [t.name for t in tl.pending()]
assert pending_names == ["write code", "read paper", "email advisor"]
tl.tasks[2].mark_done()
pending_names = [t.name for t in tl.pending()]
assert pending_names == ["write code", "read paper"]
print("B3 passed!")
```

### B4 — SimpleAgent

```python
class SimpleAgent:
    def __init__(self, name, task_list):
        self.name = name
        self.task_list = task_list

    def act(self):
        pending = self.task_list.pending()
        if not pending:
            return f"{self.name}: nothing to do!"
        task = pending[0]
        task.mark_done()
        return f"{self.name}: working on '{task.name}'"

tl = TaskList()
tl.add(Task("write code", 1))
tl.add(Task("read paper", 2))
agent = SimpleAgent("Agent Smith", tl)

r1 = agent.act()
assert r1 == "Agent Smith: working on 'write code'", r1
r2 = agent.act()
assert r2 == "Agent Smith: working on 'read paper'", r2
r3 = agent.act()
assert r3 == "Agent Smith: nothing to do!", r3
print("B4 passed!")
```
