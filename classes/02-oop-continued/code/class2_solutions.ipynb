## 🟢 Core - Coffee Shop ☕
Create 2 classes - a CoffeeShop class and a CoffeeOrder class

**CoffeeOrder**
1. Instance attributes 
- customer_name, 
- drink_name, 
- size ("small", "medium", "large"),
- price.
  
1. Instance method
- get_order_summary() - returns a string like "Alice ordered a large latte for 20 dkk"

**CoffeeShop**
1. Instance attributes
- shop_name, 
- orders (list to store CoffeeOrder objects)
2. Instance methods
- add_order(order) - adds a CoffeeOrder to the orders list
- get_total_orders() - returns how many orders the shop has
- list_all_orders() - prints all order summaries

```python
class CoffeeOrder:
    def __init__(self, customer_name: str, drink_name: str, size: str, price: float):
        self.customer_name = customer_name
        self.drink_name = drink_name
        self.price = price

        if size not in ["Small", "Medium", "Large"]:
            raise ValueError("Size must be 'Small', 'Medium', or 'Large'")
        self.size = size

    def get_order_summary(self):
        return f"{self.customer_name} ordered a {self.size} {self.drink_name} for {self.price:.2f}dkk"
```

```python
class CoffeeShop:
    def __init__(self, shop_name: str):
        self.shop_name = shop_name
        self.orders = []

    def add_order(self, order: CoffeeOrder):
        self.orders.append(order)
        
    def get_total_orders(self):
        return len(self.orders)

    def list_all_orders(self):
        for order in self.orders:
            print(order.get_order_summary())
```

```python
# create some coffee orders
c1 = CoffeeOrder("Silvia", "Latte", "Medium", 35.0)
c2 = CoffeeOrder("Julia", "Cappuccino", "Large", 47.0)

c3 = CoffeeOrder("Paul", "Chai Latte", "Small", 37.0)
c4 = CoffeeOrder("Jari", "Chocolate", "Large", 42.0)

# summary of some orders
for coffee_order in [c1, c4]:
    print(coffee_order.get_order_summary())
```

```python
# the coffee shop perspective
shop_agent = CoffeeShop(shop_name="ReDi Coffee Shop")

# adding the orders to the coffee shop
[shop_agent.add_order(coffee_order) for coffee_order in [c1, c2, c3, c4]]
```

```python
shop_agent.get_total_orders()
```

```python
shop_agent.list_all_orders()
```

### Addition - Comparison operators.

For the CoffeeOrder class, implement at least one comparison operator. The logic is up to you (e.g. compare based on price, size or both).

```python
class CoffeeOrder:
    def __init__(self, customer_name: str, drink_name: str, size: str, price: float):
        self.customer_name = customer_name
        self.drink_name = drink_name
        self.price = price

        if size not in ["Small", "Medium", "Large"]:
            raise ValueError("Size must be 'Small', 'Medium', or 'Large'")
        self.size = size

    def get_order_summary(self):
        return f"{self.customer_name} ordered a {self.size} {self.drink_name} for {self.price:.2f}dkk"
    
    def __le__(self, other):
        if not isinstance(other, CoffeeOrder):
            return NotImplemented
        return self.price <= other.price
```

## 🟢 Core — Exercise 1: The `Book` class

```python
class Book:
    def __init__(self, title, author, price, stock):
        self.title = title          # public
        self.author = author        # public
        self.__price = price        # private
        self._stock = stock         # protected

    def get_price(self):
        return self.__price

    def get_stock(self):
        return self._stock

    def sell(self):
        if self._stock > 0:
            self._stock -= 1
            return True
        return False

    def __str__(self):
        return f"{self.title} by {self.author} - {self.get_price()} dkk ({self.get_stock()} in stock)"

    def __eq__(self, other):
        return self.get_price() == other.get_price()

    def __lt__(self, other):
        return self.get_price() < other.get_price()
```

```python
b = Book("Dune", "Frank Herbert", 150, 3)
print(b)
print(b.get_price())
print(b.sell())
print(b.get_stock())

cheap = Book("Cheap Book", "Author A", 50, 5)
pricey = Book("Pricey Book", "Author B", 300, 2)
print(cheap < pricey)
print(cheap == pricey)
```

## 🟡 Stretch — Exercise 2: `EBook` subclass

```python
class EBook(Book):
    def __init__(self, title, author, price, stock, file_size_mb):
        super().__init__(title, author, price, stock)
        self.file_size_mb = file_size_mb

    def __str__(self):
        return super().__str__() + f" [ebook, {self.file_size_mb}MB]"
```

```python
e = EBook("Project Hail Mary", "Andy Weir", 120, 10, 4.2)
print(e)
```

## 🟡 Stretch / 🔴 Challenge — Exercise 3: `Library` class (full version)

```python
class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        self.books.append(book)

    def total_value(self):
        return sum(book.get_price() * book.get_stock() for book in self.books)

    def find_by_author(self, author):
        return [book for book in self.books if book.author == author]

    def __len__(self):
        return len(self.books)

    def __getitem__(self, index):
        return self.books[index]

    # ---- Challenge methods ----

    def __iter__(self):
        self._iter_index = 0
        return self

    def __next__(self):
        if self._iter_index >= len(self.books):
            raise StopIteration
        book = self.books[self._iter_index]
        self._iter_index += 1
        return book

    def save_to_file(self, filename):
        with open(filename, "w") as f:
            for book in self.books:
                f.write(f"{book.title},{book.author},{book.get_price()},{book.get_stock()}\n")

    @classmethod
    def load_from_file(cls, filename, name="Loaded Library"):
        library = cls(name)
        with open(filename, "r") as f:
            for line in f:
                title, author, price, stock = line.strip().split(",")
                library.add_book(Book(title, author, float(price), int(stock)))
        return library

    def __add__(self, other):
        merged = Library(f"{self.name} + {other.name}")
        merged.books = self.books + other.books
        return merged
```

```python
lib = Library("City Library")
lib.add_book(Book("Dune", "Frank Herbert", 150, 3))
lib.add_book(EBook("Project Hail Mary", "Andy Weir", 120, 10, 4.2))

print(len(lib))
print(lib[0])
print(lib.total_value())
print(lib.find_by_author("Andy Weir"))
```
